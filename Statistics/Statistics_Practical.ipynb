#Measure of Dispersion

```python
#range
import numpy as np

# Sample Data
arr = [1, 2, 3, 4, 5]

# Finding Max
Maximum = max(arr)
# Finding Min
Minimum = min(arr)

# Difference Of Max and Min
Range = Maximum-Minimum
print("Maximum = {}, Minimum = {} and Range = {}".format(
    Maximum, Minimum, Range))
```

Variance


```python
import statistics

arr = [1,2,4,4,6]

print("variance = ",(statistics.variance(arr)))
```

Standard deviation


```python
import numpy as np

arr = [1, 2, 3, 4, 5]
print('SD = ', (statistics.stdev(arr)))
```
