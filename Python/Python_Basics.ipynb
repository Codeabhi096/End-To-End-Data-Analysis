### Semantics of Variable Assignment in Python

When we talk about the **semantics** of variable assignment, we are talking about **what the assignment actually means or does**, rather than just its syntax.

For example:

```python
x = 10
```

The **syntax** is:

> `variable = value`

The **semantics** are:

> Python creates or identifies the object `10` and makes the name `x` refer to that object.

### Important: Python variables are names, not boxes

A common beginner misconception is:

```text
x = 10
```

means:

```text
x → [10]
```

as if `x` is a box containing 10.

A better mental model is:

```text
        ┌───────┐
x ─────►│  10   │
        └───────┘
         object
```

`x` is a **name/reference** associated with the object `10`.

---

### What happens when you reassign?

```python
x = 10
x = 20
```

After the first statement:

```text
x ───► 10
```

After the second:

```text
x ───► 20
```

The name `x` is now associated with `20`.

The assignment **doesn't change the number 10 into 20**. It changes what `x` refers to.

---

### Assignment between variables

Consider:

```python
x = 10
y = x
```

The semantics are:

```text
x ──┐
    ├──► 10
y ──┘
```

Both names refer to the same integer object.

This is important because Python's assignment generally **does not copy the object**.

---

### Example with mutable objects

This becomes especially important with lists:

```python
a = [1, 2, 3]
b = a
```

Now:

```text
a ──┐
    ├──► [1, 2, 3]
b ──┘
```

If you do:

```python
b.append(4)
```

you get:

```python
print(a)
```

Output:

```text
[1, 2, 3, 4]
```

Why?

Because `a` and `b` refer to the **same list object**.

---

### Assignment is different from comparison

Don't confuse:

```python
x = 10
```

with:

```python
x == 10
```

`=` means **assignment**:

> Make `x` refer to `10`.

`==` means **comparison**:

> Check whether `x` and `10` have equal values.

---



> **Variable assignment in Python binds a name to an object. It does not mean that the variable itself is a container holding the value. Multiple names can refer to the same object.**

So the key concept is:

**Assignment → Name → Object**

rather than:

**Variable → Box → Value**.
