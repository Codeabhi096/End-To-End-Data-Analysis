
# Data Types in Python

A **data type** tells Python **what kind of data an object represents** and therefore what operations can be performed on it.

For example:

```python
age = 25
name = "Abhishek"
price = 99.50
is_logged_in = True
```

Python understands:

```text
25       → integer
"Abhishek" → string
99.50    → floating-point number
True     → boolean
```

---

# 1. Main Built-in Data Types

Python's commonly used built-in data types can be grouped like this:

| Category | Data Types                         |
| -------- | ---------------------------------- |
| Numeric  | `int`, `float`, `complex`          |
| Boolean  | `bool`                             |
| Text     | `str`                              |
| Sequence | `list`, `tuple`, `range`           |
| Set      | `set`, `frozenset`                 |
| Mapping  | `dict`                             |
| Binary   | `bytes`, `bytearray`, `memoryview` |
| None     | `NoneType`                         |

For Data Analytics and ML, the most important ones initially are:

```text
int
float
bool
str
list
tuple
set
dict
None
```

Let's understand each.

---

# 2. `int` — Integer

`int` represents **whole numbers**, without a decimal point.

```python
age = 25
salary = 14500
count = -10
```

All are integers.

You can check the type using:

```python
age = 25

print(type(age))
```

Output:

```text
<class 'int'>
```

### Operations

```python
a = 10
b = 3

print(a + b)   # 13
print(a - b)   # 7
print(a * b)   # 30
print(a // b)  # 3
print(a % b)   # 1
print(a ** b)  # 1000
```

### Important

Python integers can become very large:

```python
x = 999999999999999999999999999999999
```

Python doesn't have the same fixed integer-size limitation you may encounter in languages such as C/C++.

---

# 3. `float` — Floating-Point Number

`float` represents numbers containing a decimal/fractional component.

```python
price = 99.50
height = 5.8
temperature = -2.5
```

```python
print(type(price))
```

Output:

```text
<class 'float'>
```

### Example

```python
x = 10.5
y = 2.5

print(x + y)
print(x * y)
```

### Scientific notation

Python also supports scientific notation:

```python
x = 2.5e3
```

This means:

```text
2.5 × 10³ = 2500.0
```

---

# 4. `complex` — Complex Numbers

Python supports complex numbers.

```python
x = 3 + 4j
```

Here:

```text
3 → real part
4 → imaginary part
j → imaginary unit
```

You can access them:

```python
x = 3 + 4j

print(x.real)
print(x.imag)
```

Output:

```text
3.0
4.0
```

You'll encounter this much less often in typical Data Analytics work, but it's part of Python's numeric types.

---

# 5. `bool` — Boolean

Boolean represents one of two values:

```python
True
False
```

Example:

```python
is_logged_in = True
is_admin = False
```

Check:

```python
print(type(is_logged_in))
```

Output:

```text
<class 'bool'>
```

Booleans are extremely important in:

* `if` statements
* filtering data
* comparisons
* loops
* Pandas operations
* ML preprocessing

Example:

```python
age = 25

print(age >= 18)
```

Output:

```text
True
```

---

# 6. `str` — String

A string represents **text**.

```python
name = "Abhishek"
city = "Patna"
```

You can use:

```python
"Hello"
'Hello'
```

Both are strings.

```python
name = "Python"

print(type(name))
```

Output:

```text
<class 'str'>
```

### Strings are sequences

For example:

```python
word = "Python"
```

Internally you can think of it as:

```text
P y t h o n
0 1 2 3 4 5
```

Therefore:

```python
print(word[0])
```

gives:

```text
P
```

And:

```python
print(word[-1])
```

gives:

```text
n
```

### Important property

Strings are **immutable**.

You cannot directly change an individual character:

```python
word = "Python"

word[0] = "J"
```

This produces an error.

Instead, you create a new string.

---

# 7. `list`

A list stores **multiple objects in an ordered collection**.

```python
numbers = [10, 20, 30, 40]
```

A list can contain different types:

```python
data = [10, "Python", 3.14, True]
```

Lists are:

* ordered
* mutable
* indexed
* allow duplicates

### Indexing

```python
numbers = [10, 20, 30, 40]

print(numbers[0])
print(numbers[2])
```

Output:

```text
10
30
```

### Modification

Because lists are mutable:

```python
numbers[0] = 100
```

Now:

```python
print(numbers)
```

Output:

```text
[100, 20, 30, 40]
```

### Adding elements

```python
numbers.append(50)
```

### Removing

```python
numbers.remove(20)
```

Lists are extremely important in Python programming and are used constantly when working with data.

---

# 8. `tuple`

A tuple is similar to a list but is **immutable**.

```python
coordinates = (10, 20)
```

You can access values:

```python
print(coordinates[0])
```

But you cannot modify them:

```python
coordinates[0] = 100
```

This produces an error.

### List vs Tuple

```text
List
[10, 20, 30]
↓
Mutable

Tuple
(10, 20, 30)
↓
Immutable
```

A tuple is useful when you want a collection whose contents should not be changed.

---

# 9. `range`

`range` represents a sequence of numbers, commonly used with loops.

```python
numbers = range(5)
```

Conceptually:

```text
0 1 2 3 4
```

Example:

```python
for i in range(5):
    print(i)
```

Output:

```text
0
1
2
3
4
```

You can also specify start and stop:

```python
range(2, 10)
```

produces:

```text
2 3 4 5 6 7 8 9
```

Notice that the ending value `10` isn't included.

---

# 10. `set`

A set stores **unique values**.

```python
numbers = {10, 20, 30, 10, 20}

print(numbers)
```

The duplicates are removed.

Conceptually:

```text
{10, 20, 30}
```

Sets are useful when you need to determine unique values.

For example:

```python
cities = {"Delhi", "Patna", "Delhi", "Mumbai"}

print(cities)
```

You get only unique cities.

Sets don't use normal positional indexing like lists.

---

# 11. `dict` — Dictionary

This is **extremely important for backend development and Data Analytics**.

A dictionary stores data as:

```text
key → value
```

Example:

```python
student = {
    "name": "Rahul",
    "age": 22,
    "marks": 85
}
```

Think of it as:

```text
"name"  → "Rahul"
"age"   → 22
"marks" → 85
```

Access a value:

```python
print(student["name"])
```

Output:

```text
Rahul
```

Modify:

```python
student["marks"] = 90
```

Add:

```python
student["city"] = "Patna"
```

Dictionaries are heavily used in:

* APIs
* JSON
* FastAPI
* databases
* configuration
* data processing

For example, an API response might look like:

```python
{
    "name": "Abhishek",
    "age": 25,
    "skills": ["Python", "SQL", "ML"]
}
```

# 12. `NoneType` — `None`

`None` represents the **absence of a value**.

```python
result = None
```

Its type is:

```python
print(type(result))
```

Output:

```text
<class 'NoneType'>
```

For example:

```python
user = None
```

This could mean:

> There is currently no user assigned.

Don't confuse:

```text
None
0
False
""
```

They are different values.

# 13. Mutable vs Immutable

This is a **very important Python concept**.

### Mutable

An object can be changed after creation.

Examples:

```text
list
dict
set
bytearray
```

Example:

```python
numbers = [1, 2, 3]

numbers.append(4)
```

The same list object has been modified.

### Immutable

The object cannot be changed after creation.

Examples:

```text
int
float
bool
str
tuple
complex
```

For example:

```python
x = 10
```

You cannot modify the integer object `10`.

When you do:

```python
x = 20
```

you're making `x` refer to another object.

This connects directly to what we discussed earlier about **variable assignment semantics**.

# 14. Checking Data Types

The most common way is:

```python
type()
```

Example:

```python
x = 100

print(type(x))
```

Output:

```text
<class 'int'>
```

You can also use `isinstance()`:

```python
x = 100

print(isinstance(x, int))
```

Output:

```text
True
```

For programming, `isinstance()` is often preferable when you want to check whether an object belongs to a particular type/class.

# 15. Type Conversion

Python allows you to convert between many types.

### String → Integer

```python
x = "100"

x = int(x)

print(x)
print(type(x))
```

Now:

```text
100
<class 'int'>
```

### Integer → Float

```python
x = 10

y = float(x)
```

Result:

```text
10.0
```

### Integer → String

```python
x = 100

y = str(x)
```

Now:

```text
"100"
```

### List → Set

```python
numbers = [1, 2, 2, 3, 3]

unique = set(numbers)
```

Result:

```text
{1, 2, 3}
```

# 16. One Important Concept: Python is Dynamically Typed

Consider:

```python
x = 10
```

Here `x` refers to an integer.

Later:

```python
x = "Python"
```

Now `x` refers to a string.

You don't have to declare:

```python
int x
```

or:

```python
string x
```

Python determines the type at runtime.

That's why Python is called **dynamically typed**.

Think:

```text
x = 10
     ↓
    int

x = "Python"
     ↓
    str
```

The **name `x` isn't permanently an integer variable**.

# 17. A Very Important Mental Model

Connect this with your previous question about assignment.

When you write:

```python
x = 10
```

think:

```text
       ┌──────────┐
x ────►│  10      │
       │ int      │
       └──────────┘
```

When you write:

```python
x = "Python"
```

think:

```text
       ┌──────────┐
x ────►│ "Python" │
       │ str      │
       └──────────┘
```

So:

> **Python variables are names that reference objects, and objects have types.**

This is a much better understanding than thinking that a variable is simply a box containing a value.
